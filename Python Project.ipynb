# SOCIAL MEDIA ENGAGEMENT ANALYSIS – SUMMARY

This project analyzed social media engagement data using Python, Pandas, NumPy, Matplotlib, Seaborn, and Plotly. The analysis focused on content performance, user behavior, and sentiment patterns.

## Key Findings

• **Content Performance:** Video posts achieved the highest average engagement rate among all post types, followed by text, image, and reel posts.

• **Content Category:** Food was the best-performing content category, followed by tech and lifestyle.

• **Country Performance:** Brazil recorded the highest average engagement rate among the countries analyzed.

• **Age and Engagement:** Engagement varied across different age groups, showing differences in user interaction patterns.

• **Verified Accounts:** Verified accounts achieved a higher average engagement rate than non-verified accounts.

• **Best Time for Impressions:** Posts published around **12 AM** recorded the highest average impressions.

• **Device Impact:** Mobile users had the highest average watch time compared with tablet and desktop users.

• **Sentiment Performance:** Negative sentiment posts recorded the highest average engagement rate, followed by neutral and positive posts.

• **Negative vs Neutral Posts:** Negative posts had higher engagement rate, likes, shares, and watch time than neutral posts. Neutral posts received slightly more comments.

## Conclusion

The analysis shows that engagement depends on content type, category, country, posting time, device usage, account verification, and sentiment. Video and food-related content performed strongly, while mobile users showed the highest watch time. These insights can help improve social media content strategy and audience engagement.
